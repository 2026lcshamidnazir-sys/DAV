```python

# ============================================================
# HANDLING MISSING VALUES USING PANDAS
# ============================================================

# Import required libraries
import pandas as pd
import numpy as np


# ============================================================
# STEP 1: CREATE A STUDENT DATASET WITH MISSING VALUES
# ============================================================

df = pd.DataFrame({

    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108],

    "Student_Name": [
        "Hamid", "Rahul", "Aman", "Priya",
        "Sara", "Arjun", "Neha", "Karan"
    ],

    "Age": [20, 21, np.nan, 22, 20, 21, 23, np.nan],

    "Study_Hours": [5, 6, 4, np.nan, 7, 5, 8, 6],

    "Attendance": [85, 90, 75, 80, np.nan, 95, 88, 92],

    "Marks": [78, 85, 65, 72, 90, np.nan, 88, 82],

    "Department": [
        "CSE", "ECE", "CSE", np.nan,
        "ISE", "CSE", "ECE", "CSE"
    ]
})


# Display original dataset
print("========== ORIGINAL STUDENT DATASET ==========")
print(df)


# ============================================================
# STEP 2: IDENTIFY MISSING VALUES
# ============================================================

print("\n========== MISSING VALUES IN EACH COLUMN ==========")

print(df.isnull().sum())


# ============================================================
# STEP 3: DROP ROWS CONTAINING MISSING VALUES
# ============================================================

# Remove rows containing one or more missing values
df_dropped = df.dropna()

print("\n========== DATASET AFTER DROPPING MISSING ROWS ==========")

print(df_dropped)


# ============================================================
# STEP 4: MEAN IMPUTATION
# ============================================================

# Replace missing Age values with the mean age
df["Age"] = df["Age"].fillna(df["Age"].mean())

print("\n========== AGE AFTER MEAN IMPUTATION ==========")

print(df["Age"])


# ============================================================
# STEP 5: MEDIAN IMPUTATION
# ============================================================

# Replace missing Study_Hours with the median
df["Study_Hours"] = df["Study_Hours"].fillna(
    df["Study_Hours"].median()
)

print("\n========== STUDY HOURS AFTER MEDIAN IMPUTATION ==========")

print(df["Study_Hours"])


# ============================================================
# STEP 6: MODE IMPUTATION
# ============================================================

# Replace missing Department with the most frequent department
df["Department"] = df["Department"].fillna(
    df["Department"].mode()[0]
)

print("\n========== DEPARTMENT AFTER MODE IMPUTATION ==========")

print(df["Department"])


# ============================================================
# STEP 7: FORWARD FILL
# ============================================================

# Create a series of student attendance readings
attendance_readings = pd.Series(
    [85, 90, np.nan, 80, np.nan, 95]
)

print("\n========== ATTENDANCE BEFORE FORWARD FILL ==========")

print(attendance_readings)

# Replace missing values with the previous available value
attendance_readings = attendance_readings.ffill()

print("\n========== ATTENDANCE AFTER FORWARD FILL ==========")

print(attendance_readings)


# ============================================================
# STEP 8: VERIFY THE FINAL DATASET
# ============================================================

# Replace missing Attendance values with median
df["Attendance"] = df["Attendance"].fillna(
    df["Attendance"].median()
)

# Replace missing Marks values with mean
df["Marks"] = df["Marks"].fillna(
    df["Marks"].mean()
)

print("\n========== FINAL STUDENT DATASET ==========")

print(df)

# Check remaining missing values
print("\n========== MISSING VALUES AFTER IMPUTATION ==========")

print(df.isnull().sum())

# Display final dataset information
print("\n========== FINAL DATASET INFORMATION ==========")

df.info()

print("\nMissing Value Handling Experiment Completed Successfully!")
```
