```python

# ============================================================
# NUMPY AND PANDAS OPERATIONS ON STUDENT DATASET
# ============================================================

# Import required libraries
import numpy as np
import pandas as pd


# ============================================================
# 1. LOAD STUDENT DATASET
# ============================================================

# Load the CSV file
data_path = "student.csv"
df = pd.read_csv(data_path)

# Remove unnecessary index columns if they exist
df = df.drop(
    columns=["Unnamed: 0", "Id"],
    errors="ignore"
)

print("========== STUDENT DATASET LOADED SUCCESSFULLY ==========")


# ============================================================
# 2. NUMPY ARRAY
# ============================================================

# Convert Weekly Study Hours column into a NumPy array
study_hours = pd.to_numeric(
    df["Weekly_Study_Hours"],
    errors="coerce"
).to_numpy()

# Remove missing values for numerical calculations
study_hours_clean = study_hours[~np.isnan(study_hours)]


print("\n========== NUMPY ARRAY ==========")

# Display NumPy array
print("Weekly Study Hours:")
print(study_hours)

# Calculate mean study hours
print("\nMean Study Hours:")
print(np.mean(study_hours_clean))

# Find maximum study hours
print("\nMaximum Study Hours:")
print(np.max(study_hours_clean))

# Find minimum study hours
print("\nMinimum Study Hours:")
print(np.min(study_hours_clean))

# Display first four study-hour values
print("\nFirst four study-hour values:")
print(study_hours[:4])

# Display last three study-hour values
print("\nLast three study-hour values:")
print(study_hours[-3:])


# ============================================================
# 3. PANDAS DATAFRAME
# ============================================================

print("\n========== PANDAS DATAFRAME ==========")

# Display complete dataset
print(df)


# ============================================================
# 4. DISPLAY FIRST FOUR ROWS
# ============================================================

print("\n========== FIRST FOUR ROWS ==========")

print(df.iloc[:4])


# ============================================================
# 5. DISPLAY STUDENT AGE COLUMN
# ============================================================

print("\n========== STUDENT AGE COLUMN ==========")

print(df["Student_Age"])


# ============================================================
# 6. FILTER STUDENTS STUDYING MORE THAN 6 HOURS
# ============================================================

print("\n========== STUDENTS WITH MORE THAN 6 WEEKLY STUDY HOURS ==========")

# Convert study hours into numeric values for filtering
weekly_hours = pd.to_numeric(
    df["Weekly_Study_Hours"],
    errors="coerce"
)

# Filter students studying more than 6 hours
print(df[weekly_hours > 6])


# ============================================================
# 7. DISPLAY FIRST THREE ROWS AND THREE COLUMNS
# ============================================================

print("\n========== FIRST THREE ROWS AND THREE COLUMNS ==========")

print(df.iloc[:3, :3])


# ============================================================
# 8. DATASET INFORMATION
# ============================================================

print("\n========== DATASET INFORMATION ==========")

df.info()


# ============================================================
# 9. DESCRIPTIVE STATISTICS
# ============================================================

print("\n========== DESCRIPTIVE STATISTICS ==========")

print(df.describe())


# ============================================================
# END OF EXPERIMENT
# ============================================================

print("\nNumPy and Pandas Operations Completed Successfully!")

```
