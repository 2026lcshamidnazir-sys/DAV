# Data Visualization Techniques

Visualization of the student dataset using Matplotlib and Seaborn.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("darkgrid")

# Load student dataset
df = pd.read_csv("student.csv").drop_duplicates().dropna()

print("Rows used:", len(df))

# Create two histograms
fig, ax = plt.subplots(1, 2, figsize=(11, 4))

# Histogram 1: Student Age
ax[0].hist(df["Student_Age"], bins=6, edgecolor="white")
ax[0].set_title("Distribution of Student Age")
ax[0].set_xlabel("Age")
ax[0].set_ylabel("Number of Students")

# Histogram 2: Weekly Study Hours
sns.histplot(
    df["Weekly_Study_Hours"],
    bins=7,
    kde=True,
    ax=ax[1]
)
ax[1].set_title("Distribution of Weekly Study Hours")
ax[1].set_xlabel("Weekly Study Hours")
ax[1].set_ylabel("Number of Students")

plt.tight_layout()
plt.show()
```
